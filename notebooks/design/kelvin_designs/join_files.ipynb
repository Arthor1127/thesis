Resonators plus Hall bar

```python
# ------------------------------------------------------------
# Merge multiple GDS files into a single mask layout
# FIXED: unique cell names per imported file
# ------------------------------------------------------------

import klayout.db as db
import os

# ------------------------------------------------------------
# Initialize layout and top-level cell
# ------------------------------------------------------------
ly = db.Layout()
top_cell = ly.create_cell("mask")

# ------------------------------------------------------------
# GDS files to import
# ------------------------------------------------------------
files = [
    "../Eight/f_8.1_to_8.5_n_8_Qc_100K.gds",
    "gds_output/hall_bar.gds",
]

# ------------------------------------------------------------
# Placement coordinates
# ------------------------------------------------------------
placements = [
    [(0, 0), (3950, 400)],
    [(0, 0), (5480, 2520)],
]

# ------------------------------------------------------------
# Import + place function (SAFE)
# ------------------------------------------------------------
def import_and_place(layout, top_cell, gds_file, position):
    ly_import = db.Layout()
    ly_import.read(gds_file)

    imported_top = ly_import.top_cell()
    if imported_top is None:
        raise RuntimeError(f"No top cell in {gds_file}")

    # ---- CRITICAL FIX: unique cell name ----
    base_name = os.path.splitext(os.path.basename(gds_file))[0]
    unique_name = f"{imported_top.name}_{base_name}"

    target_cell = layout.create_cell(unique_name)
    target_cell.copy_tree(imported_top)

    # Place instance
    x, y = position
    inst = db.DCellInstArray(
        target_cell.cell_index(),
        db.DTrans(db.DVector(x, y))
    )
    top_cell.insert(inst)

# ------------------------------------------------------------
# Perform placements
# ------------------------------------------------------------
for placement_set in placements:
    for gds_file, (x, y) in zip(files, placement_set):
        import_and_place(ly, top_cell, gds_file, (x, y))

# ------------------------------------------------------------
# Write output GDS
# ------------------------------------------------------------
output_file = "gds_output/res_Nb_grAl_hall_bar_v1.gds"
ly.write(output_file)

print(f"Mask written to: {output_file}")
```

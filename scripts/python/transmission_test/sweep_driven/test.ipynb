```python
import numpy as np
import matplotlib.pyplot as plt
import scienceplots

plt.style.use(['science', 'no-latex', 'notebook'])
```

```python
pos_1 = np.load('./pos_3.95270/s_sweep.npz')
pos_3 = np.load('./pos_4.17770//s_sweep.npz')
pos_2 = np.load('./pos_4.10270//s_sweep.npz')
```

```python
pos_1.files
```

```python
freqs = pos_1['freq_hz']
```

```python
s21_1 = pos_1['S21']
s21_2 = pos_2['S21']
s21_3 = pos_3['S21']
```

```python
x_1 = 3.95270
x_2 = 4.10270
x_3 = 4.17770
```

```python
plt.plot(freqs/ 1e9, 20 * np.log(np.abs(s21_1)), label = fr'$x =$ {x_1:.2f} mm', linewidth = 4)
plt.plot(freqs/ 1e9, 20 * np.log(np.abs(s21_2)), label = fr'$x =$ {x_2:.2f} mm', linewidth = 4)
plt.plot(freqs/ 1e9, 20 * np.log(np.abs(s21_3)), label = fr'$x =$ {x_3:.2f} mm', linewidth = 4)
plt.xlabel('Frequency [GHz]')
plt.ylabel(r'$\vert S_{21} \vert$ [dB]')
plt.legend()
# plt.yscale('log')
```

```python
plt.plot(freqs, np.angle(s21_1))
plt.plot(freqs, np.angle(s21_2))
plt.plot(freqs, np.angle(s21_3))
```
